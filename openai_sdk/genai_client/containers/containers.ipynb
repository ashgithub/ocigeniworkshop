# Containers Practice Notebook

What this notebook does:

1. Builds the configured OpenAI client for OCI endpoint usage.
2. Creates, lists, and retrieves hosted containers.
3. Uploads, lists, retrieves, and downloads container files.
4. Cleans up test resources only when `DELETE_AT_END=True`.

Use this notebook when you want to run the same beginner flow cell by cell.

## 1. Setup

Before running:
1. Configure OCI credentials in `sandbox.yaml`.
2. Install project dependencies (`uv sync`).
3. Use the included sample input file, or replace `LOCAL_FILE_PATH` with your own file.

```python
# Step 1: Build a configured OpenAI client for OCI endpoint usage.
from openai import OpenAI
from openai_sdk.openai_client_provider import OpenAIClientProvider

print('<------------- Step 1: Configure OpenAI client ------------>')
client: OpenAI = OpenAIClientProvider().oci_openai_client
print('Client ready')
```

```python
# Fill these values for your workspace if you want to reuse existing resources.
CONTAINER_ID = ''
FILE_ID = ''
LOCAL_FILE_PATH = 'openai_sdk/genai_client/containers/sample_container_input.md'
# Keep this False while learning so deletion only happens when you choose it.
DELETE_AT_END = False
```

## 2. Container Lifecycle

These cells show the same core operations as `base_example.py`: create a container, list containers, and retrieve one by id.

```python
# Step 2a: Create a hosted container for this notebook run.
print('<------------- Step 2a: Create container ------------>')
created_container = client.containers.create(name='workshop-container')
print('Created:', created_container.id)
CONTAINER_ID = CONTAINER_ID or created_container.id
```

```python
# Step 2b: List available containers and print how many records came back.
print('<------------- Step 2b: List containers ------------>')
containers_page = client.containers.list()
print('Containers returned:', len(containers_page.data))
```

```python
# Step 2c: Retrieve one container by id to confirm the resource can be fetched.
print('<------------- Step 2c: Retrieve container ------------>')
if not CONTAINER_ID:
    raise ValueError('Set CONTAINER_ID before retrieve')
container = client.containers.retrieve(CONTAINER_ID)
container
```

## 3. Container File Operations

These cells show the same shape as `container_files.py`: upload a local file, inspect file metadata, then retrieve the file bytes.

```python
# Step 3a: Upload a local file into the selected container.
print('<------------- Step 3a: Upload file ------------>')
with open(LOCAL_FILE_PATH, 'rb') as fh:
    uploaded = client.containers.files.create(container_id=CONTAINER_ID, file=fh)
print('Uploaded file:', uploaded.id)
FILE_ID = FILE_ID or uploaded.id
```

```python
# Step 3b: List files currently attached to this container.
print('<------------- Step 3b: List container files ------------>')
files_page = client.containers.files.list(container_id=CONTAINER_ID)
print('Files in container:', len(files_page.data))
```

```python
# Step 3c: Retrieve file metadata. This does not download the file bytes.
print('<------------- Step 3c: Retrieve file metadata ------------>')
if not FILE_ID:
    raise ValueError('Set FILE_ID before retrieve/content')
file_meta = client.containers.files.retrieve(container_id=CONTAINER_ID, file_id=FILE_ID)
file_meta
```

```python
# Step 3d: Retrieve and read the file content stream.
print('<------------- Step 3d: Retrieve file content ------------>')
content = client.containers.files.content.retrieve(container_id=CONTAINER_ID, file_id=FILE_ID)
bytes_data = content.read()
print('Downloaded bytes:', len(bytes_data))
```

## 4. Optional Cleanup

Cleanup is intentionally gated by `DELETE_AT_END`. Set it to `True` only when you are ready to delete the uploaded file and container.

```python
# Step 4: Delete resources only when the cleanup flag is enabled.
print('<------------- Step 4: Optional cleanup ------------>')
if DELETE_AT_END and FILE_ID:
    print(client.containers.files.delete(container_id=CONTAINER_ID, file_id=FILE_ID))
if DELETE_AT_END and CONTAINER_ID:
    print(client.containers.delete(CONTAINER_ID))
if not DELETE_AT_END:
    print('Skipping cleanup. Set DELETE_AT_END=True to delete resources.')
```
